# Laboratorium 17

Do wykresu z poprzedniego laboratorium dodano węzły:

* **Thief Caught** - szanse na złapanie złodzieja.



* **Action** - blok decyzyjny o podejmowanej przez Holmesa kroku.

*Nic nie rób*, *Jedź do domu*, *Dzwoń na policję*.

* **Payoff** - blok użyteczności, przedstawiający zysk lub stratę w kontekście podjętych kroków, ustawione w skali -100 do 2000.

1. Holmes otrzymuje telefon od Watsona, nic nie słyszy w radiu o trzęsieniu. Podejmuje decyzję o tym, że sam pojedzie do domu. Niestety nie łapie złodzieja w pojedynkę:
