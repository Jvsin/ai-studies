## Laboratorium 13:

### Przykład 1:

Problem medyczny:

Sieć bayesowska jest przydatna w problemach medycznych, pozwala na poprawne decyzje diagnostyczne oraz interpretację testów.

### Przykład 2

Problem roślinny:

Przy rolnictwie również ma sens, pozwala na diagnostykę problemów uprawy, planowanie interwencji oraz badanie wpływów różnych czynników.

### Przykład 3

Problem ekologiczny:

W tym temacie rodzi się problem, gdyż niektóre elementy mogą powodować siebie nawzajem, np. przekształcanie lasów w pustynie -> zwrost gazów cieplarnianych (CO2) oraz wzrost temperatury. 

Co za tym idzie, pojawiają się **cykle**, co uniemożliwia użycia sieci Bayesowskiej.
