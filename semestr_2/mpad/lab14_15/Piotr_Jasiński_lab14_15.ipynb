# Laboratorium 14 i 15:

## Zadanie 14.1

### Połączenie szeregowe:

Jeśli nie znamy stanu **Zapalenie płuc**, informacja przepływa (Grypa zwiększa prawdobieństwo X-ray). 

Jeśli zaobserwujemy **Zapalenie płuc**, dodatkowa wiedza o grypie nie zmienia już naszej wiedzy o X-ray (bo X-ray zależy bezpośrednio tylko od stanu płuc).

### Połączenie rozbieżne:

**Grypa jest wspólną przyczyną**.

**Jeśli nie wiemy, czy pacjent ma grypę**: kaszel i zapalenie płuc są zależne (masz kaszel: pewnie masz grypę, która zwiększa nam prawdopodobieństwo zapalenia płuc). 

**Jeśli wiemy na pewno, że pacjent ma grypę**: kaszel staje się niezależny od zapalenia płuc (warunkowa niezależność) - skoro ma grypę to prawdopodobieństwo zapalenia płuc jest takie same, niezależnie od tego czy pacjent kaszle czy nie.

### Połączenia zbieżne:

**Nie wiemy, czy pacjent kaszle**: Jeśli nie mamy żadnych informacji o objawach pacjenta, to fakt, że zachorował on na Grypę, nie zmienia prawdopodobieństwa, że mógł też zachorować na COVID. Te dwa zdarzenia są niezależne od siebie. Złapanie jednej choroby nie sprawia, że prawdopodobieństwo drugiej się zmienia.

**Wiemy, że pacjent kaszle:** Teraz Grypa i COVID stają się zależne warunkowo. Obie choroby są potencjalnymi wyjaśnieniami tego samego obserwowanego faktu.

## Zadanie 15:

Węzeł Grypa jest przyczyną kaszlu jak i kichania. Jeśli zaobserwujemy Grypę, występowanie jednego symptomu nie wpływa na szanse drugiego.

**KONSULTACJA LEKARZA**: Objawy mogą być powiązane nawet przy znanej diagnozie. Może występować inna przyczyna, która powoduje zarówno kaszel jak i kichanie. Przykładem może być **alergia**.

**Wiemy, że pacjent ma grypę, ale nie wiemy, czy pacjent ma alergię:** Skoro pacjent nie ma grypy, kaszel może być spowodowany przez alergię, która powoduje też kichanie. Zatem kichanie nadal wpływa na prawdopodobieństwo kaszlu, przez węzeł alergii, której jeszcze nie wiemy.
