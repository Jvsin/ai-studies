## Parametry wejściowe:

### Burglary: 

50 włamań w ciągu 2 lat (730 dni) w sąsiedztwie 1350 domów. 
P(Burglary) = 50 / (730 * 1350) = 0.00005
* **P(True) =** 0.00005
* **P(False) =** 0.99995



### Earthquake:
Zakładamy średnio 3 trzęsienia ziemi rocznie. P(Earthquake) = 3 / 365 = 0.0082

* **P(True) =** 0.0082
* **P(False) =** 0.9918



### Alarm:
* **Burglary = True | Earthquake = False**: 
Z zadania - Alarm działa w **95%** skuteczności. -> P(True) = 0.95.
* **Burglary = True | Earthquake = True**:
Można założyć że zadziała z nieco większym prawdopodobieństwem niż bez trzęsienia ziemi - **97%**. -> P(True) = 0.97.

* **Burglary = False | Earthquake = True**:
Z zadania - Alarm włącza się przy 1 na 5 trzęsień ziemi, **20%** szans. -> P(True) = 0.2
* **Burglary = False | Earthquake = False**:
Nic się nie dzieje, więc szansa uruchomienia powinna być minimalna - błąd systemu, np. **0.0001%**. -> P(True) = 0.0001.





### Radio News:
Z zadania - nigdy nie zawiedli informując poprawnie w aktualnym stanie, lecz potrafią podać informację gdy nastąpi w innym stanie.
* **Earthquake = True:** P(True) = 1.
* **Earthquake = False:** P(True) = 0.02 (czasem się pomylą, ale nie aż tak często).



### Watson Call:

* **Alarm = True:** Watson jest troche głuchy, więc może nie usłyszeć: **P(True) = 0.95**.
* **Alarm = False:** Watson nie żartuje, jest poważny: **P(True) = 0**.
