### Gęstość prawdopodobieństwa

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 400)
std_devs = [0.2, 0.5, 1, 1.5, 2]
mean = 0

for std in std_devs:
    y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / std) ** 2)
    plt.plot(x, y, label=f'std={std}')

plt.xlabel('x')
plt.ylabel('probability density')
plt.title('gaussian plot')
plt.legend()
plt.show()
```

### Dystrybuanta prawdopodobieństwa

```python
x = np.linspace(-5, 5, 400)
std_devs = [0.2, 0.5, 1, 1.5, 2]
mean = 0

for std in std_devs:
    y = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / std) ** 2)
    cdf = np.cumsum(y)
    cdf = cdf / cdf[-1]
    plt.plot(x, cdf, label=f'std={std}')

plt.xlabel('x')
plt.ylabel('cumulative density')
plt.title('gaussian plot')
plt.legend()
plt.show()
```

```python
fig, axs = plt.subplots(2, 1, figsize=(8, 8), sharex=True)

for std in std_devs:
    y_pdf = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / std) ** 2)
    axs[0].plot(x, y_pdf, label=f'std={std}')
axs[0].set_ylabel('Probability Density')
axs[0].set_title('Gęstość prawdopodobieństwa')
axs[0].legend()

for std in std_devs:
    y_pdf = (1 / (std * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mean) / std) ** 2)
    cdf_val = np.cumsum(y_pdf)
    cdf_val = cdf_val / cdf_val[-1]
    axs[1].plot(x, cdf_val, label=f'std={std}')
axs[1].set_xlabel('x')
axs[1].set_ylabel('Cumulative Probability')
axs[1].set_title('dystrybuanta prawdopodobieństwa')
axs[1].legend()

plt.tight_layout()
plt.show()
```

### Rozkład masy prawdopodobieństwa

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 100
p1 = 0.5
p2 = 0.1

k = np.arange(0, n+1)

pmf1 = binom.pmf(k, n, p1)
pmf2 = binom.pmf(k, n, p2)

plt.figure(figsize=(10, 6))
plt.plot(k, pmf1, 'ro', label='Binom dla 0.5', markersize=10)  
plt.plot(k, pmf2, 'gx', label='Binom dla 0.1', markersize=10) 

plt.title('rozkład masy prawdop.')
plt.xlabel('liczba sukcesów')
plt.ylabel('prawdopodobieństwo')

plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
